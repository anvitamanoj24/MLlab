```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_excel(r"Online Retail.xlsx")
print(df.shape)
print(df.head(5))
```

```python
print(df.isnull().sum())
```

```python
df=df.dropna(subset=['CustomerID'])
df=df[(df['Quantity']>0)&(df['UnitPrice'])>0]
df['TotalSpend']=df['Quantity']*df['UnitPrice']
print(df.shape)
print(df.head(5))
```

```python
customer_df=df.groupby('CustomerID').agg({
    "Quantity":"sum",
    "UnitPrice":"sum",
    "TotalSpend":"sum",
    "Country":"first"
}).reset_index()
```

```python
threshold=customer_df['TotalSpend'].median()
customer_df['HighValue']=(customer_df['TotalSpend']>threshold).astype(int)
print(customer_df.head(5))
print(customer_df.shape)
```

```python
customer_df['Country']=LabelEncoder().fit_transform(customer_df['Country'])
```

```python
X=customer_df[['Quantity','UnitPrice','Country']]
y=customer_df['HighValue']
```

```python
model=DecisionTreeClassifier(criterion='entropy',max_depth=4)
model.fit(X,y)
```

```python
plt.figure(figsize=(20,8))
plot_tree(model,
feature_names=X.columns,
class_names=['Low','High'],
filled=True
)
plt.title('Customer Segmentation using decision tree')
plt.show()
```
